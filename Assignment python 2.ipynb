```python
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}
print(my_set)
```

Explanation:
A set is an unordered collection of unique elements in Python.
Duplicate elements are automatically removed from a set.
In the given list, the element 'a' appears three times and 'i' appears twice.
Python stores only one occurrence of each element.
The order of elements in a set is not fixed, so the output may appear in a different order.
The output contians only five unique vowels: 'a', 'e', 'i', 'o', 'u'.

```python
my_set[4] = 's'
```

Explanation: 
Sets are mutable , meaning elements can be added or removed. 
However, sets do not support indexing.
Therefore, we cannot access or modify an element using my_set[4].
This statement produces a TypeError.

```python
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}
print("Set 1:",set1)
print("Set 2:",set2)
```

```python
print("Union:", set1.union(set2))
print("Intersection: ",set1.intersection(set2))
```
